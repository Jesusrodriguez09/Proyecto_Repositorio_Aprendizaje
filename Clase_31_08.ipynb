```python
import matplotlib.pyplot as plt
import numpy as np
```

# **Dos Gráficas;**

```python
x = 0.4 + np.arange(8)
y = [4.8, 5.5, 3.5, 4.6, 6.5, 6.6, 2.6, 3.0]

plt.bar(x, y, width=1, edgecolor="white", linewidth=0.7, color="purple")
plt.xlim=(0, 8)
plt.xticks=np.arange(1, 8),
plt.ylim=(0, 8)
plt.yticks=np.arange(1, 8)
plt.show()
```

```python
np.random.seed(1)
x = 4 + np.random.normal(0, 1.5, 200)

plt.hist(x, bins=8, linewidth=0.5, edgecolor="white", color = "green")
plt.title("Un Histograma")
plt.xlabel("Distribución")
plt.ylabel("Frecuencia")
plt.xlim=(0, 8)
plt.xticks=np.arange(1, 8)
plt.ylim=(0, 56)
plt.yticks=np.linspace(0, 56, 9)
plt.show()
```

# **Gráficas Realizadas Durante La Clase**;

```python
x = np.linspace(0,9,10)
print(x)
```

```python
from matplotlib.lines import lineStyles
y = x**2

plt.plot(x,y, color = "green", marker ="H", linestyle=":")
plt.title("$f(x)=x^2$")
plt.ylabel("$x^2$")
plt.xlabel("$x$")

plt.show()
```

```python
s = np.linspace(0,20,100)
t = np.sin(s)
plt.plot(s,t)
plt.title("$F(x) = Sen(x)$")
plt.xlabel("$x$")
plt.ylabel("$Sen(x)$")
plt.grid()
plt.show()
```

```python
s = np.linspace(0,20,100)
t = np.sin(s)
plt.plot(s,t)
plt.title("$F(x) = Sen(x)$")
plt.xlabel("$x$")
plt.ylabel("$Sen(x)$")
plt.xlim(0, 30) #dominio
plt.grid()
plt.show()
```

```python
x = np.linspace(0,10,100)
y =  np.sin(x)
y2 = np.cos(x)


plt.plot(x,y)
plt.plot(x,y2)
plt.title("$Sen(x), Cos(x)$")
plt.xlabel("x")
plt.grid()
```

```python
from matplotlib import legend
#Subplots

plt.subplot( 2, 1, 1) #filas, #columnas,
plt.plot(x,y, label="$Sen(x)$")
plt.title("$Sen(x)$")
plt.xlabel("$x$")
plt.legend()
plt.grid()



plt.subplot(2, 1, 2)
plt.plot(x,y2, color = "red", label="$Cos(x)$")
plt.title("$Cos(x)$")
plt.xlabel("$x$")
plt.legend()
plt.grid()

plt.tight_layout()
```
